# Day 2 — Data Quality and Relationship Analysis

## Objective

Evaluate the quality and consistency of the Olist datasets by analyzing:

- Missing values
- Duplicate records
- Key uniqueness
- Referential integrity
- Relationships between tables

These checks help ensure that the data is reliable before building the demand forecasting and inventory optimization pipeline.

## 1. Import Required Libraries

Import Pandas for data manipulation and NumPy for numerical operations.

```python
import pandas as pd
import numpy as np
import os
```

```python
# ## 2. Define Project Data Path

# Define the location of the raw Olist datasets stored in Google Drive.
PROJECT_PATH = "/content/drive/MyDrive/Projects/AI_Smart_Warehouse_Management/"

RAW_DATA_PATH = os.path.join(
    PROJECT_PATH,
    "01_Raw Data"
)

print(RAW_DATA_PATH)
```

```python
from google.colab import drive
drive.mount('/content/drive')

# ## 3. Load Olist Datasets

# Load the nine Olist tables required for data quality assessment and relationship validation.
orders = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_orders_dataset.csv")
)

order_items = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_order_items_dataset.csv")
)

products = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_products_dataset.csv")
)

customers = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_customers_dataset.csv")
)

sellers = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_sellers_dataset.csv")
)

payments = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_order_payments_dataset.csv")
)

reviews = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_order_reviews_dataset.csv")
)

geolocation = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "olist_geolocation_dataset.csv")
)

category_translation = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "product_category_name_translation.csv")
)

print("All Olist datasets loaded successfully.")
```

```python
## 4. Organize Datasets

# Create a dictionary containing all datasets so that common data-quality checks can be performed efficiently.
datasets = {
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Customers": customers,
    "Sellers": sellers,
    "Payments": payments,
    "Reviews": reviews,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}
```

```python
## 5. Dataset Dimensions

# Check the number of rows and columns in each dataset to understand the volume of available data.
for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
## 6. Missing Value Analysis

#Identify #missing values in each dataset. Missing-value analysis helps determine which
#fields require further investigation before modeling.
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
## 7. Missing Value Percentage

#Calculate the percentage of missing values in each column to understand the severity of missing data.
for name, df in datasets.items():
    missing_pct = (df.isnull().sum() / len(df) * 100)
    missing_pct = missing_pct[missing_pct > 0].sort_values(ascending=False)

    print(f"\n{name}")
    print(missing_pct)
```

```python
## 8. Duplicate Record Analysis

# Check for completely duplicated rows in each dataset to identify potential duplicate records that could affect analysis.
for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    print(f"{name}: {duplicate_count} duplicate rows")
```

```python
## 9. Primary Key Validation

#Check whether the main identifier columns are unique. This helps confirm the structure of the relational Olist database.
print("ORDERS")
print("Rows:", len(orders))
print("Unique order IDs:", orders["order_id"].nunique())

print("\nPRODUCTS")
print("Rows:", len(products))
print("Unique product IDs:", products["product_id"].nunique())

print("\nCUSTOMERS")
print("Rows:", len(customers))
print("Unique customer IDs:", customers["customer_id"].nunique())

print("\nSELLERS")
print("Rows:", len(sellers))
print("Unique seller IDs:", sellers["seller_id"].nunique())
```

```python
## 10. Referential Integrity — Orders

#Verify that every order referenced in the Order Items table exists in the Orders table.
missing_orders = set(order_items["order_id"]) - set(orders["order_id"])

print("Orders referenced in Order Items:",
      order_items["order_id"].nunique())

print("Orders missing from Orders table:",
      len(missing_orders))
```

```python
## 11. Referential Integrity — Products

#Verify that every product referenced in Order Items exists in the Products table.
missing_products = set(order_items["product_id"]) - set(products["product_id"])

print("Products referenced in Order Items:",
      order_items["product_id"].nunique())

print("Products missing from Products table:",
      len(missing_products))
```

```python
## 12. Referential Integrity — Sellers

#Verify that every seller referenced in Order Items exists in the Sellers table.
missing_sellers = set(order_items["seller_id"]) - set(sellers["seller_id"])

print("Sellers referenced in Order Items:",
      order_items["seller_id"].nunique())

print("Sellers missing from Sellers table:",
      len(missing_sellers))
```

```python
## 13. Referential Integrity — Customers

#Verify that every customer associated with an order exists in the Customers table.
missing_customers = set(orders["customer_id"]) - set(customers["customer_id"])

print("Customers referenced in Orders:",
      orders["customer_id"].nunique())

print("Customers missing from Customers table:",
      len(missing_customers))
```

```python
## 14. Product Category Validation

# Check the relationship between the product category names in the Products table and their English translations.
product_categories = set(
    products["product_category_name"].dropna()
)

translated_categories = set(
    category_translation["product_category_name"].dropna()
)

missing_translations = product_categories - translated_categories

print("Product categories:", len(product_categories))
print("Translated categories:", len(translated_categories))
print("Categories without translation:", len(missing_translations))
```
