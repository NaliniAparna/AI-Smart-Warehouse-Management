```python
# ### Load Supporting Olist Tables
# Load customer, seller, payment, review, geolocation, and product-category translation data to understand the complete e-commerce ecosystem.
customers = pd.read_csv(
    os.path.join(project_path, "olist_customers_dataset.csv")
)

sellers = pd.read_csv(
    os.path.join(project_path, "olist_sellers_dataset.csv")
)

payments = pd.read_csv(
    os.path.join(project_path, "olist_order_payments_dataset.csv")
)

reviews = pd.read_csv(
    os.path.join(project_path, "olist_order_reviews_dataset.csv")
)

geolocation = pd.read_csv(
    os.path.join(project_path, "olist_geolocation_dataset.csv")
)

category_translation = pd.read_csv(
    os.path.join(project_path, "product_category_name_translation.csv")
)
```
